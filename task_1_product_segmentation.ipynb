```python
df_clean=df.drop_duplicates(subset="product_id").copy()
df_clean.shape
```

```python
df_clean['discounted_price'] = (
    df_clean['discounted_price']
    .str.replace('₹', '', regex=False)
    .str.replace(',', '', regex=False)
)

df_clean['discount_percentage'] = (
    df_clean['discount_percentage']
    .str.replace('%', '', regex=False)
)

df_clean['rating_count'] = (
    df_clean['rating_count']
    .str.replace(',', '', regex=False)
)
```

```python
cols = [
    'discounted_price',
    'discount_percentage',
    'rating',
    'rating_count'
]

for col in cols:
    df_clean[col] = pd.to_numeric(
        df_clean[col],
        errors='coerce'
    )

df_clean[cols].dtypes
```

```python
df_clean[cols].isnull().sum()
```

```python
df_clean[df_clean[cols].isnull().any(axis=1)]
```

```python
df_clean = df_clean.dropna(subset=cols).copy()

df_clean.shape
```

```python
from sklearn.preprocessing import StandardScaler

# Select clustering features
X = df_clean[cols].copy()

# Reduce skewness in price and rating count
X['discounted_price'] = np.log1p(X['discounted_price'])
X['rating_count'] = np.log1p(X['rating_count'])

# Apply scaling
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print(X_scaled.shape)
```

```python
from sklearn.cluster import KMeans

inertia = []

K = range(2, 9)

for k in K:
    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    model.fit(X_scaled)
    inertia.append(model.inertia_)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8, 5))

plt.plot(K, inertia, marker='o')

plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.title('Elbow Method for K-Means')

plt.xticks(list(K))
plt.grid(True)

plt.show()
```

```python
from sklearn.metrics import silhouette_score
from sklearn.cluster import KMeans

silhouette_scores = []

for k in range(2, 9):

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X_scaled)

    score = silhouette_score(X_scaled, labels)

    silhouette_scores.append(score)

    print(f"K={k}, Silhouette Score={score:.4f}")
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    range(2, 9),
    silhouette_scores,
    marker='o'
)

plt.xlabel('Number of Clusters (K)')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Score Analysis')

plt.xticks(range(2, 9))
plt.grid(True)

plt.show()
```

```python
best_k = np.argmax(silhouette_scores) + 2

print("Best number of clusters:", best_k)
print("Best Silhouette Score:", max(silhouette_scores))
```

```python
kmeans = KMeans(
    n_clusters=5,
    random_state=42,
    n_init=10
)

df_clean['Cluster'] = kmeans.fit_predict(X_scaled)

df_clean[['product_id', 'Cluster']].head(10)
```

```python
df_clean['Cluster'].value_counts().sort_index()
```

```python
cluster_summary = df_clean.groupby('Cluster')[cols].median()

cluster_summary
```

```python
import seaborn as sns
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df_clean,
    x='discounted_price',
    y='rating_count',
    hue='Cluster',
    palette='tab10'
)

plt.title('Amazon Product Segmentation')
plt.xlabel('Discounted Price')
plt.ylabel('Rating Count')
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df_clean,
    x='discount_percentage',
    y='rating',
    hue='Cluster',
    palette='tab10'
)

plt.title('Discount vs Rating by Cluster')
plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering

hierarchical = AgglomerativeClustering(
    n_clusters=5,
    linkage='ward'
)

df_clean['Hierarchical_Cluster'] = (
    hierarchical.fit_predict(X_scaled)
)

df_clean[
    ['product_id', 'Cluster', 'Hierarchical_Cluster']
].head(10)
```

```python
from sklearn.metrics import silhouette_score

kmeans_score = silhouette_score(
    X_scaled,
    df_clean['Cluster']
)

hierarchical_score = silhouette_score(
    X_scaled,
    df_clean['Hierarchical_Cluster']
)

print("K-Means Score:", round(kmeans_score, 4))
print("Hierarchical Score:", round(hierarchical_score, 4))
```

```python
from sklearn.metrics import adjusted_rand_score

ari = adjusted_rand_score(
    df_clean['Cluster'],
    df_clean['Hierarchical_Cluster']
)

print("Adjusted Rand Index:", round(ari, 4))
```

```python
cluster_summary = df_clean.groupby('Cluster').agg(
    Products=('product_id', 'count'),
    Median_Price=('discounted_price', 'median'),
    Median_Discount=('discount_percentage', 'median'),
    Median_Rating=('rating', 'median'),
    Median_Rating_Count=('rating_count', 'median')
)

cluster_summary.round(2)
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

features = [
    'Median_Price',
    'Median_Discount',
    'Median_Rating',
    'Median_Rating_Count'
]

titles = [
    'Median Discounted Price',
    'Median Discount Percentage',
    'Median Customer Rating',
    'Median Rating Count'
]

for ax, feature, title in zip(axes.flat, features, titles):
    sns.barplot(
        x=cluster_summary.index,
        y=cluster_summary[feature],
        ax=ax,
        color='steelblue'
    )
    ax.set_title(title)
    ax.set_xlabel('Cluster')

plt.tight_layout()
plt.show()
```

## Task 1: Product Segmentation — Final Conclusion

The objective of this analysis was to segment Amazon products based on discounted price, discount percentage, customer rating, and rating count using unsupervised machine learning techniques.

After data cleaning and preprocessing, **1,348 unique products** were selected for clustering. Log transformation and StandardScaler were applied to prepare the features for K-Means Clustering.

The Elbow Method and Silhouette Score were used to determine the appropriate number of clusters. The analysis identified **five product segments**, representing different pricing levels, discount strategies, customer satisfaction levels, and popularity patterns.

### Key Findings

1. **Cluster 0 — Low-Rated Products:** Products with relatively low customer ratings and limited engagement, indicating potential areas for product improvement.

2. **Cluster 1 — Popular and Highly Discounted Products:** Products with strong customer engagement, a median rating of 4.20, a median rating count of 20,850, and a median discount of 60%.

3. **Cluster 2 — Premium Products:** Higher-priced products with a median discounted price of ₹8,799 and generally positive customer ratings.

4. **Cluster 3 — Budget-Friendly Products:** Affordable products with high discounts but comparatively lower customer engagement.

5. **Cluster 4 — Moderately Priced Products:** Products with relatively low discounts, good customer ratings, and moderate customer engagement.

### Clustering Model Comparison

K-Means Clustering achieved a **Silhouette Score of 0.236**, compared with **0.1613 for Hierarchical Clustering**. This indicates that K-Means produced comparatively better-defined clusters. However, the relatively low silhouette scores suggest some overlap between product segments.

### Strongest Commercial Potential

**Cluster 1 was identified as the segment with the strongest commercial potential** because it combines affordable pricing, high discounts, positive customer ratings, and the highest median rating count among the five clusters.

The high rating count suggests strong customer engagement and potential market demand, making this segment a promising candidate for marketing and promotional strategies.

### Limitations

The dataset does not contain actual sales, revenue, or profit information. Therefore, customer ratings and rating counts were used as indirect indicators of market demand. The results cannot establish actual profitability or prove that discounts cause higher demand.

### Overall Conclusion

K-Means Clustering successfully provided a useful segmentation of Amazon products into five groups based on pricing, discounts, ratings, and customer engagement.

The analysis identified Cluster 1 as the most promising segment based on the available demand indicators. These insights can support data-driven decisions related to product positioning, marketing, pricing, and customer satisfaction.

Further analysis using actual sales and profitability data would help validate the commercial potential of each segment.
