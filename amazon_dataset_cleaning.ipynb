```python

import pandas as pd
import numpy as np
import re


df = pd.read_csv("amazon.csv")

print("Original Dataset Shape:", df.shape)

duplicates = df["product_id"].duplicated().sum()
print("Duplicate Product IDs:", duplicates)

df = df.drop_duplicates(
    subset="product_id",
    keep="first"
).copy()



for col in ["discounted_price", "actual_price"]:
    df[col] = (
        df[col]
        .astype("string")
        .str.replace(r"[₹,\s]", "", regex=True)
    )
    df[col] = pd.to_numeric(df[col], errors="coerce")


df["discount_percentage"] = (
    df["discount_percentage"]
    .astype("string")
    .str.replace("%", "", regex=False)
)

df["discount_percentage"] = pd.to_numeric(
    df["discount_percentage"],
    errors="coerce"
)

# Convert rating and rating_count
df["rating"] = pd.to_numeric(
    df["rating"], errors="coerce"
)

df["rating_count"] = pd.to_numeric(
    df["rating_count"]
    .astype("string")
    .str.replace(",", "", regex=False),
    errors="coerce"
)



text_columns = [
    "product_name",
    "category",
    "about_product",
    "review_title",
    "review_content"
]

for col in text_columns:
    df[col] = (
        df[col]
        .astype("string")
        .str.replace(r"\s+", " ", regex=True)
        .str.strip()
    )



print("\nMissing Values Before Cleaning:")
print(df.isnull().sum())

essential_columns = [
    "product_id",
    "product_name",
    "category",
    "discounted_price",
    "actual_price",
    "discount_percentage",
    "rating",
    "rating_count"
]

df = df.dropna(subset=essential_columns)

df = df.dropna(
    subset=["review_content", "review_title"]
)

df = df[
    df["rating"].between(1, 5)
    & (df["rating_count"] >= 0)
    & (df["discounted_price"] > 0)
    & (df["actual_price"] > 0)
    & df["discount_percentage"].between(0, 100)
].copy()

df = df.reset_index(drop=True)

print("\nFinal Dataset Shape:", df.shape)

print("\nData Types:")
print(df.dtypes)

print("\nRemaining Missing Values:")
print(df.isnull().sum())

print("\nRemaining Duplicate Product IDs:")
print(df["product_id"].duplicated().sum())

print("\nNumerical Statistics:")
display(
    df[
        [
            "discounted_price",
            "actual_price",
            "discount_percentage",
            "rating",
            "rating_count"
        ]
    ].describe().round(2)
)

df.to_csv(
    "amazon_cleaned.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\nAmazon Cleaned CSV Exported Successfully!")
print("Filename: amazon_cleaned.csv")
print("Total Products:", len(df))
print("Total Columns:", len(df.columns))

display(df.head())
```
